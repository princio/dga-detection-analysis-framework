```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine(f"postgresql+psycopg2://postgres:@localhost:5432/dns_mac")

def query(day: int, window_length: int, idxwindow: int):
    return pd.read_sql(f"""
SELECT
	CASE WHEN DAC_RANK=1 THEN DAC_FAMILY ELSE 'HEALTHY' END AS DAC_FAMILY_2,

	COUNT(*) AS QR,
	COUNT(*) FILTER (WHERE RCODE IS NULL) AS Q,

	COUNT(*) FILTER (WHERE RCODE=0 AND DN_RANK = 1 AND EPS1 >= 0.5) AS OK,
	COUNT(*) FILTER (WHERE RCODE=3 AND DN_RANK = 1 AND EPS1 >= 0.5) AS NX,
	
	COUNT(*) FILTER (WHERE RN_QR_RCODE=1 AND RCODE=0 AND DN_RANK = 1 AND EPS1 >= 0.5) AS UOK,
	COUNT(*) FILTER (WHERE RN_QR_RCODE=1 AND RCODE=3 AND DN_RANK = 1 AND EPS1 >= 0.5) AS UNX
    
FROM
   GET_WINDOW_ALL2({day}, {window_length}, {idxwindow})
GROUP BY DAC_FAMILY_2
""", engine)
```

```python

dfs = []
for idxday in range(10):
    for idxwindow in range(24):
        print(idxday, idxwindow)
        df = query(idxday, 3600, idxwindow)
        dfs.append(df)
        pass
    pass

df = pd.concat(dfs)
```

```python
import random
import pandas as pd
from sklearn.utils import shuffle

# Definizione delle variabili
total_elements = 12
train_size = 8
test_size = 4

# Creare un dataset bilanciato (6 elementi di classe 0 e 6 di classe 1)
dataset = [0 for i in range(total_elements // 2)] + [1 for i in range(total_elements // 2, total_elements)]

# Mescolare casualmente il dataset per evitare ordinamenti predefiniti
random.shuffle(dataset)
dataset = [ (i,c) for i,c in enumerate(dataset)]

# Funzione per suddividere il dataset mantenendo il bilanciamento tra training e testing
def balanced_split(dataset, train_size, test_size):
    # Separare gli elementi per classe
    class_0 = [item for item in dataset if item[1] == 0]
    class_1 = [item for item in dataset if item[1] == 1]

    # Determinare quanti elementi prendere per il training e per il testing per ciascuna classe
    train_size_per_class = train_size // 2
    test_size_per_class = test_size // 2

    # Prendere elementi per il training e per il testing mantenendo il bilanciamento
    train_0 = class_0[:train_size_per_class]
    test_0 = class_0[train_size_per_class:train_size_per_class + test_size_per_class]
    
    train_1 = class_1[:train_size_per_class]
    test_1 = class_1[train_size_per_class:train_size_per_class + test_size_per_class]

    # Combinare i set di training e testing bilanciati
    training_set = train_0 + train_1
    testing_set = test_0 + test_1

    # Mescolare i set per avere elementi ordinati casualmente
    random.shuffle(training_set)
    random.shuffle(testing_set)

    return training_set, testing_set

# Eseguire la suddivisione bilanciata
train_set, test_set = balanced_split(dataset, train_size, test_size)

# Convertire i set di training e testing in dataframe per visualizzarli
train_df = pd.DataFrame(train_set, columns=["Elemento", "Classe"])
test_df = pd.DataFrame(test_set, columns=["Elemento", "Classe"])

for row in train_df.to_numpy().tolist():
    n  = row[0]
    cl = row[1]
    print(f'{n}/{cl}', end=',')
    pass
print()

for i, row in test_df.iterrows():
    n  = row['Elemento']
    cl = row['Classe']
    print(f'{n}/{cl}', end=',')
    pass
print()

for i, row in dataset:
    n  = row[0]
    cl = row[1]
    print(f'{n}/{cl}', end=',')
```
